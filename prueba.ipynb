```python
import requests



class MoodleAPI:
    def __init__(self, api_url, token):
        self.api_url = api_url
        self.token = token

    def update_grade(self, course_id, component, activity_id, student_id, grade, feedback):
        function_name = 'core_grades_update_grades'
        params = {
            "wstoken": self.token,
            "wsfunction": function_name,
            "moodlewsrestformat": "json",
            "source": "custom_source",
            "courseid": course_id,
            "component": component,
            "activityid": activity_id,
            "itemnumber": 0,
            "grades[0][studentid]": student_id,
            "grades[0][grade]": grade,
            "grades[0][str_feedback]": feedback
        }

        response = requests.post(self.api_url, data=params, verify=False)
        return self._handle_response(response)

    def get_user_by_field(self, field, value):
        function_name = 'core_user_get_users_by_field'
        params = {
            'wstoken': self.token,
            'wsfunction': function_name,
            'moodlewsrestformat': 'json',
            'field': field,
            'values[0]': value
        }
        
        response = requests.get(self.api_url, params=params, verify=False)
        return self._handle_response(response)

    def _handle_response(self, response):
        if response.status_code == 200:
            return response.json()
        return {
            "error": True,
            "status_code": response.status_code,
            "message": response.text
        }

# Example
api_url = 'https://formacion.ghips.co/moodle/webservice/rest/server.php'

# Token de usuario con permisos para modificar calificaciones

token= "a6b62135f99e445ac5b38b6c9187f6cb"

moodle = MoodleAPI(api_url, token)

# Obtener el id de un usuario por su email

try:
    user = moodle.get_user_by_field('email', 'alejandro.hernandeza@udea.edu.co')
    #user_id = user['users'][0]['id']
    #print(user_id)
    print(user)
except Exception as e:
    print (e)   




```
